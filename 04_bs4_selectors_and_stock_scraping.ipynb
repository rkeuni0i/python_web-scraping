```python

df = pd.DataFrame()
for page in range(1,2) :
    stock_url = "http://finance.naver.com/item/sise_day.naver?code=005930" + "&page=" + str(page)
    headers = {"User-Agent" : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36'}
    res = requests.get(stock_url, headers)

    new_data = pd.read_html(res.text, header=0)[0]

    df = pd.concat([df, new_data], ignore_index=True)

df = df.dropna()
df
```
